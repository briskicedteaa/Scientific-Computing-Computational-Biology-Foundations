```python
!pip install biopython
from Bio import SeqIO
from Bio.SeqUtils.ProtParam import ProteinAnalysis
```

```python
for record in SeqIO.parse("P14416.txt", "swiss"):
  print(f"Sequence: {record.seq}")
  print(f"sequence length: {len(record)}")
  print(f"Protein Name: {record.description}")
  print(f"Organism: {record.annotations['organism']}")
  print(f"Gene Name: {record.annotations['gene_name']}")

D2R_organism = record.annotations['organism']
D2R_gene = record.annotations['gene_name'][0]['Name']
D2R_name = record.description
D2R_seq = (record.seq)
D2R_seq_length = len(record)
```

```python
D2Rseq = D2R_seq

analysed_D2Rseq = ProteinAnalysis(D2Rseq)
D2R_MolWt = analysed_D2Rseq.molecular_weight()

print(f"D2R MolWt: {D2R_MolWt}")
```

```python
D2R_aa_counts = analysed_D2Rseq.count_amino_acids()

most_frequent_D2R_aa = max(D2R_aa_counts, key=D2R_aa_counts.get)
highest_D2R_count = D2R_aa_counts[most_frequent_D2R_aa]

print(f"Most frequent amino acid: {most_frequent_D2R_aa} (appears {highest_D2R_count} times)")
```
