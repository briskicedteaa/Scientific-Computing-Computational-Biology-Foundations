```python
!pip install biopython
from Bio import SeqIO
from Bio.SeqUtils.ProtParam import ProteinAnalysis
```

```python
for record in SeqIO.parse("P14416.fasta", "fasta"):
  print(f"Sequence: {record.seq}")
  print(f"sequence length: {len(record)}")

D2R_seq = (record.seq)
D2R_seq_length = {len(record)}
```

```python
D2Rseq = str(D2R_seq)

analysed_D2R_seq = ProteinAnalysis(D2R_seq)

D2R_MolWt = analysed_D2R_seq.molecular_weight()

print(f"D2R MolWt: {D2R_MolWt}")
```

```python
for record in SeqIO.parse("P22303.fasta", "fasta"):
  print(f"Sequence: {record.seq}")
  print(f"sequence length: {len(record)}")

ACES_seq = {record.seq}
ACES_seq_length = {len(record)}
```

```python
sequence_object = next(iter(ACES_seq))
ACESseq = str(sequence_object)

sequence_length_object = next(iter(ACES_seq_length))

analysed_ACES_seq = ProteinAnalysis(ACESseq)

ACES_MolWt = analysed_ACES_seq.molecular_weight()

print(f"ACES Molecular Weight: {ACES_MolWt}")
```

```python
for record in SeqIO.parse("Q16620.fasta", "fasta"):
  print(f"Sequence: {record.seq}")
  print(f"sequence length: {len(record)}")

TrkB_seq = {record.seq}
TrkB_seq_length = {len(record)}
```

```python
sequence_object = next(iter(TrkB_seq))
TrkBseq = str(sequence_object)

sequence_length_object = next(iter(TrkB_seq_length))

analysed_TrkB_seq = ProteinAnalysis(TrkBseq)

TrkB_MolWt = analysed_TrkB_seq.molecular_weight()

print(f"TrkB MolWt: {TrkB_MolWt}")
```

```python
import pandas as pd

ProteinData = {'Protein': ['D2R', 'ACES', 'TrkB'],
               'Sequence Length': [D2R_seq_length, ACES_seq_length, TrkB_seq_length],
               'Mol Wt': [D2R_MolWt, ACES_MolWt, TrkB_MolWt]
}

df = pd.DataFrame(ProteinData)

print(df)
```
